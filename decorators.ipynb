```python
# Timing function Execution

import time

def timer(fun):
    def wrapper(*args,**kwargs):
        start = time.time()
        result = fun(*args,**kwargs)
        end = time.time()
        print(f"{fun.__name__} ran in {end-start} time ")
        return result
    return wrapper


@timer
def sleep_fun(n):
    time.sleep(n)


sleep_fun(2)
```

```python
# Debugging functions call
# create a decorators to print the function name and the values of its arguments every time the function is called.

def debug(fun):
    def wrapper(*args,**kwargs):
        args_value = ' ,'.join(str(args) for arg in args )
        kwargs_value = ' ,'.join(f"{k}={v}" for k , v in kwargs.items())
        print(f"calling: {fun.__name__} with args {args_value} and kwargs {kwargs_value}")
        return fun(*args,**kwargs)

    return wrapper

@debug
def hello():
    print("hello , guys ")

@debug 
def greet(name , greeting= "Nice to meet you"):
    print(f"{greeting},{name}")

hello()
greet("William")
```

```python
# Cache Return Value 
# Implement the decorator that cache the return value of the function , so that when it called with the same arguments,the cache value is return instead of re-executing the function.

import time 

def cache(fun):
    cache_value = {}
    def wrapper(*args):
        if args in cache_value:
            return cache_value[args]
        result = fun(*args)
        return result

    return wrapper

@cache 
def long_running(a,b):
    time.sleep(2)
    return a*b

print(long_running(6,3))
print(long_running(5,4))
```
