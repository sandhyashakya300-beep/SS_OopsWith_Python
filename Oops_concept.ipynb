```python
# modify the car class to encapualte brand attribute, making it private and provide a getter mehtod for it.
#Encapsulation
class Car:
    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model

    def get_brand(self):
        return self.__brand + "!"

    def full_name(self):
        return f"{self.__brand} {self.model}"  


class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size


my_tesla = ElectricCar("Tesla", "Model s", 100)
my_car = Car("Toyota", "Corolla")
print(my_car.get_brand())
print(my_car.model)
print(my_car.full_name())
print(my_tesla.battery_size)
print(my_tesla.full_name())
print(my_tesla.get_brand())
```

```python
# Polymorphism 
# Demonstrate polymorphism by defining a method fuel type in the car and ElectricCar classes , but with different behavior.
class Car:
    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model

    def get_brand(self):
        return self.__brand + "!"

    def full_name(self):
        return f"{self.__brand} {self.model}"  

    def fuel_type(self):
        return "Petrol or Diesel"

    
class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

    def fuel_type(self):
            return " Electric Charge"

my_tesla = ElectricCar("Tesla", "Model s", 100)
my_car = Car("Toyota", "Corolla")
print(my_car.fuel_type())
print(my_tesla.fuel_type())
```

```python
# Property Decorators
# Use a Property decorators in the class to make the model attributes read only.
class Car:
    def __init__(self, brand, model):
        self.__brand = brand
        self.__model = model

    def get_brand(self):
        return self.__brand + "!"

    def full_name(self):
        return f"{self.__brand} {self.__model}"  

    def fuel_type(self):
        return "Petrol or Diesel"

    @property
    def model(self):
         return self.__model

    
class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

    def fuel_type(self):
            return " Electric Charge"

my_tesla = ElectricCar("Tesla", "Model s", 100)
my_car = Car("Toyota", "Corolla")
#my_car.model = "New model"
print(my_car.model)
print(my_car.fuel_type())
```

```python
# class inheritance and isinstance() function
# Demonstrate the use of isinstance() to check if my_tesla is an instance  of a car and Electriccar classes
class Car:
    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model

    def get_brand(self):
        return self.__brand + "!"

    def full_name(self):
        return f"{self.__brand} {self.model}"  

    def fuel_type(self):
        return "Petrol or Diesel"

    
class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

    def fuel_type(self):
            return " Electric Charge"

my_tesla = ElectricCar("Tesla", "Model s", 100)
print(isinstance(my_tesla, Car ))
print(isinstance(my_tesla, ElectricCar ))
```
